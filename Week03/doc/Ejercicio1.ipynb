# Librerías

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, recall_score, precision_score, f1_score, roc_auc_score, roc_curve, r2_score
```

# Implementación de código

```python
# Dataset:
y_true = [0, 1, 0, 0, 1, 1]
y_pred = [0, 0, 1, 0, 0, 1]
# Matriz de confusión:
cm = confusion_matrix(y_true, y_pred)

# Gráfico de la CM

plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False)
plt.xlabel('Predicciones')
plt.ylabel('Valores Verdaderos')
plt.title('Matriz de Confusión')

```

```python
# Exactitud:
print("Exactitud ", accuracy_score(y_true, y_pred))
# Sensibilidad:
print("Sensibilidad ", recall_score(y_true, y_pred))
# Especificidad:
print("Especificidad ", recall_score(y_true, y_pred, pos_label=0))
# Precisión:
print("Precisión ", precision_score(y_true, y_pred))
# Puntuación F1:
print("Puntuación F1 ", f1_score(y_true, y_pred))
# Área bajo la curva:
auc = roc_auc_score(y_true, y_pred)
print("Área bajo la curva:", auc)
# R Score:(R^2 coefficient of determination)
print("R Score", r2_score(y_true, y_pred))
# ROC Curve (binary classification problem):
plt.plot(roc_curve(y_true, y_pred)[0], roc_curve(y_true, y_pred)[1], color='darkorange',lw=2, label='ROC curve (area = %0.2f)' % auc)
```
